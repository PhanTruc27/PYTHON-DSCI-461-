# Sublime Limes' Line Graphs

- You are a data analyst for an online lime retailer called Sublime Limes. People all over the world can use this product to get the freshest, best-quality limes delivered to their door.
- One of the product managers at Sublime Limes would like to gain insight into **the customers and their sales habits**. 

```python

from matplotlib import pyplot as plt
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
visits_per_month = [9695, 7909, 10831, 12942, 12495, 16794, 14161, 12762, 12777, 12439, 10309, 8724]
```

```python
# numbers of limes of different species sold each month
key_limes_per_month = [92.0, 109.0, 124.0, 70.0, 101.0, 79.0, 106.0, 101.0, 103.0, 90.0, 102.0, 106.0]
persian_limes_per_month = [67.0, 51.0, 57.0, 54.0, 83.0, 90.0, 52.0, 63.0, 51.0, 44.0, 64.0, 78.0]
blood_limes_per_month = [75.0, 75.0, 76.0, 71.0, 74.0, 77.0, 69.0, 80.0, 63.0, 69.0, 73.0, 82.0]
```

1. Create a figure of width=12, height=8

```python
plt.figure(figsize=(12, 8))
plt.plot(months, visits_per_month, marker='o')
plt.xlabel("Months")
plt.ylabel("Visits per Month")
plt.title("FIGURE: Website Visits per Month")
plt.show()
```

2. Make two charts in one figure, laid out side-by-side

```python
fig,axes = plt.subplots(1,2,figsize = (12,8))
axes[0].plot(months, visits_per_month, marker='o')
axes[0].set_title("Website Visits per Month")
axes[0].set_xlabel("Months")
axes[0].set_ylabel("Visits")

axes[1].plot(months, key_limes_per_month, marker='o')
axes[1].set_title("Key Limes Sold per Month")
axes[1].set_xlabel("Months")
axes[1].set_ylabel("Number of Limes")

plt.tight_layout()
plt.show()
```

3. In the left subplot, we are going to plot the total page visits over the past year
   -  create the list of 'x-values', which is range(len(months))
   -  Plot the **total page visits** against these **x_values** as a line. Give the line markers that will help show each month as a distinct value.
   -  Label the x-axis and y-axis with descriptive titles of what they measure.
   -  Set the x-axis ticks to be the x_values.
   -  Label the x-axis tick labels to be the names stored in the months list.


```python
# Create x-values
x_values = range(len(months))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 8))

#LEFT 
axes[0].plot(x_values, visits_per_month, marker='o')
axes[0].set_title("Total Page Visits Over the Past Year")
axes[0].set_xlabel("Months")
axes[0].set_ylabel("Total Page Visits")
axes[0].set_xticks(x_values)
axes[0].set_xticklabels(months)

plt.tight_layout()
plt.show()
```

 4. In the subplot on the right, we are going to plot three lines on the same set of axes
    - number of key limes sold vs x_values
    - number of Persian limes sold vs x_values
    - number of blood limes sold vs x_values
      


```python

fig, axes = plt.subplots(1, 2, figsize=(12, 8))

#LEFT 
axes[0].plot(x_values, visits_per_month, marker='o')
axes[0].set_title("Total Page Visits Over the Past Year")
axes[0].set_xlabel("Months")
axes[0].set_ylabel("Total Page Visits")
axes[0].set_xticks(x_values)
axes[0].set_xticklabels(months)

# RIGHT SUBPLOT 
axes[1].plot(x_values, key_limes_per_month, marker='o', label="Key Limes")
axes[1].plot(x_values, persian_limes_per_month, marker='o', label="Persian Limes")
axes[1].plot(x_values, blood_limes_per_month, marker='o', label="Blood Limes")

axes[1].set_title("Lime Sales per Month")
axes[1].set_xlabel("Months")
axes[1].set_ylabel("Number of Limes Sold")
axes[1].set_xticks(x_values)
axes[1].set_xticklabels(months)
axes[1].legend()


plt.tight_layout()
plt.show()
```

5. Give each line a specific color of your choosing.

6. Add a legend to differentiate the lines, labeling each lime species.

7. Set the x-axis ticks to be the x_values, and the tick labels to be the months list.

```python
# ---------------- FIGURE ----------------
fig, axes = plt.subplots(1, 2, figsize=(12, 8))

# ---- LEFT SUBPLOT ----
axes[0].plot(x_values, visits_per_month, marker='o')
axes[0].set_title("Total Page Visits Over the Past Year")
axes[0].set_xlabel("Months")
axes[0].set_ylabel("Total Page Visits")
axes[0].set_xticks(x_values)
axes[0].set_xticklabels(months)

# ---- RIGHT SUBPLOT ----
axes[1].plot(x_values, key_limes_per_month,
             color='green', marker='o', label='Key Limes')

axes[1].plot(x_values, persian_limes_per_month,
             color='orange', marker='o', label='Persian Limes')

axes[1].plot(x_values, blood_limes_per_month,
             color='red', marker='o', label='Blood Limes')

axes[1].set_title("Lime Sales per Month")
axes[1].set_xlabel("Months")
axes[1].set_ylabel("Number of Limes Sold")
axes[1].set_xticks(x_values)
axes[1].set_xticklabels(months)
axes[1].legend()

# ---------------- SHOW ----------------
plt.tight_layout()
plt.show()
```
