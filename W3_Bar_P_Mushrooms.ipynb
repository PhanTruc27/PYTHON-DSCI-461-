````python
import matplotlib.pyplot as plt
import seaborn as sns

for column in df.columns:
    # Thiết lập kích thước biểu đồ cho dễ nhìn
    plt.figure(figsize=(10, 6))
    
    # Vẽ countplot (biểu đồ cột đếm tần suất)
    sns.countplot(
        x=column,
        data=df,
        order=df[column].value_counts().index,
        palette="viridis"
    )
    
    # Tiêu đề và xoay nhãn trục x
    plt.title(f"Value Counts of {column}")
    plt.xticks(rotation=45)
    
    plt.tight_layout()
    plt.show();
````

## Cleaning the Bar Graphs

6. As you scroll through the graphs, you may notice some imperfections. For example, some of the x-axis labels overlap each other. The font size for the labels along the x-axis is also pretty small, making them tough to read.

   Let's fix these up with two lines of code.
   
- rotates the value labels slightly so they don't overlap, also slightly increases font size
   plt.xticks(rotation=30, fontsize=10)
- increases the variable label font size slightly to increase readability
   plt.xlabel(column, fontsize=12)
   ```
- add an informative title. Using `.title()` from the Matplotlib library, give your graph the following title:

   ```
   {Variable Name} Value Counts
   ```

````python
import matplotlib.pyplot as plt
import seaborn as sns

for column in df.columns:
    plt.figure(figsize=(10, 6))
    
    sns.countplot(
        x=column,
        data=df,
        order=df[column].value_counts().index,
        palette="viridis"
    )
    
    # Xoay nhãn trục x và tăng font size
    plt.xticks(rotation=30, fontsize=10)
    
    # Tăng font size cho nhãn trục x
    plt.xlabel(column, fontsize=12)
    
    # Thêm title cho biểu đồ
    plt.title(f"{column} Value Counts", fontsize=14)
    
    plt.tight_layout()
    plt.show()
````

````python
import matplotlib.pyplot as plt
columns = df.columns
````

````python

for column in columns:
    col_name = str(column)   # 👈 ép tuple → string

    counts = df[column].value_counts()
    order = counts.index

    plt.figure()
    plt.bar(order, counts.loc[order].values)

    plt.xticks(rotation=30, fontsize=10)
    plt.xlabel(col_name, fontsize=12)
    plt.title(col_name + " Value Counts")

    plt.show()
    plt.clf()
````
